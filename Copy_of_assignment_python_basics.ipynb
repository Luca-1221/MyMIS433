Next, let's make it a bit challenging by adding conditional statements to convert temperatures to and from Celsius, Fahrenheit. [Formula: Celsius/5 = (Fahrenheit – 32)/9]

An example program output:
```

```

Another example:
```
Please enter the temperature: 140
Is this Celsius or Fahrenheit? F
140F is 60 in Celsius
```

````python
# complete your program here
d = input('Please enter the temperature: ')
m = input('Is this in Celsius or Fahrenheit? ')
temp = float(d)
if m == 'C':
    f = (temp * 9/5) + 32
    print(f'{temp}C is {f} in Fahrenheit')
elif m == 'F':
    c = (temp - 32) * 5/9
    print(f'{temp}F is {c} in Celsius')
else:
    print("Invalid unit. Please enter C or F.")
````

````python
# Use a loop to create a list of first n numbers in the Fibonacci sequence
# [0, 1, 1, 2, 3, 5, 8, ...]
fibonacci = []
n = int(input('How many numbers do you want? '))

# Your loop logic is perfect!
for i in range(n):
    if i == 0:
        fibonacci.append(0)
    elif i == 1:
        fibonacci.append(1)
    else:
        # Calculate the sum of the two previous elements
        next_number = fibonacci[i-1] + fibonacci[i-2]
        fibonacci.append(next_number)

print(fibonacci)
````

````python
# Count the occurrence of each alphabet (no space ' ') in this sentence.
sentence = 'the quick brown fox jumps over the lazy dog'

# Save the result in a dictionary
ch_dict = {}

# Use a loop to iterate all characters
for char in sentence:
    # 1. Check if the character is NOT a space
    if char != ' ':
        # 2. If it's already in the dictionary, add 1 to the count
        if char in ch_dict:
            ch_dict[char] += 1
        # 3. If it's the first time we see the letter, set count to 1
        else:
            ch_dict[char] = 1

# print the dictionary
print(ch_dict)
````
